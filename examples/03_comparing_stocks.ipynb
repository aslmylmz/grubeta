# Comparing Systematic Risk Across Stocks

Not all stocks carry the same market risk. **Systematic risk** (measured by beta)
tells you how much of a stock's movement is driven by the overall market versus
its own idiosyncratic factors.

By comparing betas across stocks, you can:

- Identify which holdings contribute the most market risk to your portfolio.
- Build more balanced portfolios by mixing high-beta and low-beta names.
- Understand sector-level risk patterns.

## High-Beta vs Low-Beta Stocks

Let's compare two groups:

1. **High-beta stocks** -- typically growth/tech names that move aggressively with the market.
2. **Low-beta stocks** -- typically defensive/consumer staples names that are more stable.

```python
# Group 1: High-beta growth stocks
result = compare_betas(["NVDA", "TSLA", "META"], market="SPY")
print(result["summary"])
```

## What does this mean?

Notice how growth/tech stocks (NVDA, TSLA, META) tend to have betas above 1.0 — they amplify market moves.
When the S&P 500 drops 1%, NVDA might drop 1.5% or more. But when the market rallies, these stocks rally harder too.

This is the **risk-return tradeoff**: higher beta = higher systematic risk = higher expected return over time.

```python
# Group 2: Low-beta defensive stocks
result2 = compare_betas(["JNJ", "PG", "KO"], market="SPY")
print(result2["summary"])
```

## Defensive Stocks Are Different

Consumer staples (JNJ, PG, KO) tend to have betas below 1.0. They move less than the market in both directions.
These are "defensive" stocks — they provide relative protection during market downturns.

**Key insight for portfolio construction:** Mixing high-beta and low-beta stocks lets you control your portfolio's overall market sensitivity.

```python
from grubeta import compare_betas

# All stocks together
all_stocks = ["NVDA", "TSLA", "META", "JNJ", "PG", "KO"]
result = compare_betas(all_stocks, market="SPY")
print(result["summary"])
```

## Summary

The beta comparison clearly shows two clusters:
- **High-beta cluster** (tech/growth): Betas consistently above 1.0, with more volatile trajectories
- **Low-beta cluster** (defensive): Betas consistently below 1.0, more stable over time

Dynamic beta captures how these relationships **change over time** — notice how betas shift during market stress periods (e.g., COVID crash in March 2020, rate hikes in 2022).

## Key Insight

- **High-beta stocks** (NVDA, TSLA, META) amplify market moves. When the market rallies,
  they tend to rally harder. When the market drops, they tend to fall more.
- **Low-beta stocks** (JNJ, PG, KO) dampen market moves. They provide relative stability
  during downturns but may lag during strong bull markets.

### Implications for Portfolio Construction

- **Aggressive portfolios** tilt toward high-beta names to maximize exposure to market upside.
- **Defensive portfolios** tilt toward low-beta names to reduce drawdowns.
- **Balanced portfolios** blend both to target a specific overall portfolio beta (e.g., ~1.0).

Because beta changes over time, it is important to re-estimate periodically.
A stock that was low-beta last year may have shifted -- GRUBeta captures these dynamics.
