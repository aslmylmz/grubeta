# Estimating Dynamic Beta in 3 Lines of Code

**Beta** measures how much a stock moves relative to the market.

- **Beta > 1**: The stock *amplifies* market moves. If the market goes up 1%, the stock tends to go up more than 1%.
- **Beta = 1**: The stock *tracks* the market one-for-one.
- **Beta < 1**: The stock *dampens* market moves. It is less volatile than the market.

Traditional beta is a single static number, but real-world beta changes over time.
**GRUBeta** estimates a *dynamic* beta that adapts as market conditions evolve.

## Step 1: Import and Run

That's it -- one import and one function call. GRUBeta handles data download,
model training, and visualization automatically.

```python
from grubeta import estimate_beta

result = estimate_beta("AAPL", "SPY")
```

## What does this mean?

The `result` dictionary contains everything you need:

- `result["summary"]` -- A plain-English interpretation of the estimated beta.
- `result["beta"]` -- A pandas Series of the time-varying beta values.
- `result["model"]` -- The fitted GRU model (for advanced users).

Let's start with the summary:

```python
print(result["summary"])
```

## The Beta Plot

The plot generated above shows:

- **X-axis**: Time (trading days).
- **Y-axis**: Estimated beta at each point in time.
- **The line**: How AAPL's sensitivity to SPY has evolved over the estimation window.

Notice how beta is *not* constant -- it shifts as market regimes change.

```python
# The plot was shown automatically. To access the raw data:
result["beta"].dropna().tail(10)
```
